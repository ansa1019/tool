```python
# 匯入相關套件
import os
from time import sleep
from dotenv import load_dotenv
from selenium import webdriver
from selenium.webdriver.common.by import By
from selenium.webdriver.common.keys import Keys
from selenium.webdriver.chrome.options import Options
from selenium.webdriver.support.ui import WebDriverWait
from selenium.webdriver.support import expected_conditions as EC


# region
def wait_and_click(xpath, timeout=5, use_js=False, use_click=True):
    """
    等待指定 xpath 的元素出現並點擊。

    Args:
        xpath (str): XPath 路徑
        timeout (int): 最多等待秒數
        use_js (bool): 是否用 JavaScript 點擊（預設 False）
    """
    if use_click:
        element = WebDriverWait(driver, timeout).until(
            EC.element_to_be_clickable((By.XPATH, xpath))
        )
    else:
        element = WebDriverWait(driver, timeout).until(
            EC.visibility_of_element_located((By.XPATH, xpath))
        )
    if use_js:
        driver.execute_script("arguments[0].click();", element)
    else:
        element.click()
# endregion
```

```python
# 初始設定
load_dotenv()
URL = f"https://www.instagram.com/p/DdjN9GPS40b/"
PHONE = os.getenv("PHONE")
PASSWORD = os.getenv("PASSWORD")
MAX_RETRIES = 100
comment = "想要穿去日本玩😍"
times = 10
```

```python
# 預先準備
options = Options()
driver = webdriver.Chrome(options=options)

# 檢查狀態
driver.get(URL)
WebDriverWait(driver, 3).until(
    lambda d: d.execute_script("return document.readyState") == "complete"
)
```

```python
# 登入流程
wait_and_click('//*[starts-with(@href, "/accounts/login/")]', use_js = True)
print("🔐 點擊登入按鈕")

# 輸入帳密並登入
print("📝 輸入帳號密碼登入...")
WebDriverWait(driver, 5).until(
    EC.visibility_of_element_located((By.XPATH, '//*[@name="email"]'))
)
driver.find_element(By.XPATH, '//*[@name="email"]').send_keys(PHONE)
driver.find_element(By.XPATH, '//*[@name="pass"]').send_keys(PASSWORD)
wait_and_click('//*[text()="登入"]')
print("✅ 登入")

# 等待人工驗證
input("⏳ 請完成人工驗證後按 Enter 繼續...")
```

```python
for i in range(1, times + 1):
    # 輸入留言並送出
    WebDriverWait(driver, 30).until(
        EC.element_to_be_clickable((By.XPATH, '//*[starts-with(@aria-label, "留言")]'))
    ).send_keys(comment)
    sleep(0.5)  # 等待半秒
    WebDriverWait(driver, 30).until(
        EC.element_to_be_clickable((By.XPATH, '//*[starts-with(@aria-label, "留言")]'))
    ).send_keys(Keys.ENTER)

    # 進度條
    progress = int(i / times * 20)
    bar = "█" * progress + " " * (20 - progress)
    print(f"📊 [{bar}] {i}/{times}", end="\r", flush=True)

print("\n🎉 留言完成！")
```
